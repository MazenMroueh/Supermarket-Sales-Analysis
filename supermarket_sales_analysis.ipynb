# Supermarket Sales Analysis
### Week 2 Data Analysis Portfolio Project

**Objective:** Analyze supermarket transaction data to understand sales performance, customer behavior, product performance, and branch-level results.

**Tools:** Python, Pandas, Matplotlib


## 1. Business Questions

- Which branch generates the most revenue?
- Which product line generates the most revenue?
- Which customer type contributes the most revenue?
- What is the average transaction value?
- Which payment method is most common?
- Which day of the week generates the most sales?
- How do customer ratings compare across branches?


```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (10, 5)

DATA_PATH = '../data/supermarket_sales.csv'
df = pd.read_csv(DATA_PATH)
df.head()
```

## 2. Initial Data Inspection


```python
print('Rows and columns:', df.shape)
print('\nColumns:')
print(df.columns.tolist())
```

```python
df.info()
```

```python
print('Missing values:')
print(df.isnull().sum())

print('\nDuplicate rows:', df.duplicated().sum())
```

## 3. Data Cleaning

Convert the date field to a proper datetime type and create day/month features for analysis.


```python
df['Date'] = pd.to_datetime(df['Date'])
df['Day'] = df['Date'].dt.day_name()
df['Month'] = df['Date'].dt.month_name()
df['Hour'] = pd.to_datetime(df['Time'], format='%H:%M').dt.hour

print(df[['Date', 'Day', 'Month', 'Hour']].head())
```

## 4. Key Performance Indicators (KPIs)


```python
total_revenue = df['Total'].sum()
transactions = df['Invoice ID'].nunique()
avg_transaction = df['Total'].mean()
total_quantity = df['Quantity'].sum()
avg_rating = df['Rating'].mean()

print(f'Total revenue: ${total_revenue:,.2f}')
print(f'Number of transactions: {transactions:,}')
print(f'Average transaction value: ${avg_transaction:,.2f}')
print(f'Total quantity sold: {total_quantity:,}')
print(f'Average customer rating: {avg_rating:.2f}/10')
```

## 5. Revenue by Branch


```python
branch_sales = df.groupby('Branch')['Total'].sum().sort_values(ascending=False)
print(branch_sales)

ax = branch_sales.plot(kind='bar', title='Total Revenue by Branch')
ax.set_xlabel('Branch')
ax.set_ylabel('Revenue ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('../visualizations/revenue_by_branch.png', dpi=150)
plt.show()
```

## 6. Revenue by Product Line


```python
product_sales = df.groupby('Product line')['Total'].sum().sort_values(ascending=False)
print(product_sales)

ax = product_sales.sort_values().plot(kind='barh', title='Revenue by Product Line')
ax.set_xlabel('Revenue ($)')
ax.set_ylabel('Product Line')
plt.tight_layout()
plt.savefig('../visualizations/revenue_by_product_line.png', dpi=150)
plt.show()
```

## 7. Customer Type Analysis


```python
customer_sales = df.groupby('Customer type')['Total'].sum().sort_values(ascending=False)
print(customer_sales)

ax = customer_sales.plot(kind='bar', title='Revenue by Customer Type')
ax.set_xlabel('Customer Type')
ax.set_ylabel('Revenue ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('../visualizations/revenue_by_customer_type.png', dpi=150)
plt.show()
```

## 8. Payment Method Analysis


```python
payment_counts = df['Payment'].value_counts()
print(payment_counts)

ax = payment_counts.plot(kind='bar', title='Transactions by Payment Method')
ax.set_xlabel('Payment Method')
ax.set_ylabel('Number of Transactions')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('../visualizations/payment_methods.png', dpi=150)
plt.show()
```

## 9. Sales by Day of Week


```python
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
daily_sales = df.groupby('Day')['Total'].sum().reindex(day_order)
print(daily_sales)

ax = daily_sales.plot(kind='bar', title='Revenue by Day of Week')
ax.set_xlabel('Day')
ax.set_ylabel('Revenue ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../visualizations/revenue_by_day.png', dpi=150)
plt.show()
```

## 10. Branch Ratings


```python
branch_rating = df.groupby('Branch')['Rating'].mean().sort_values(ascending=False)
print(branch_rating)

ax = branch_rating.plot(kind='bar', title='Average Customer Rating by Branch')
ax.set_xlabel('Branch')
ax.set_ylabel('Average Rating')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('../visualizations/average_rating_by_branch.png', dpi=150)
plt.show()
```

## 11. Automated Findings

The following cells turn the analysis into concise business findings. This keeps the conclusions tied to the actual data.


```python
top_branch = branch_sales.idxmax()
top_branch_revenue = branch_sales.max()
top_product = product_sales.idxmax()
top_product_revenue = product_sales.max()
top_customer_type = customer_sales.idxmax()
top_payment = payment_counts.idxmax()
top_day = daily_sales.idxmax()

print('KEY FINDINGS')
print(f'1. Highest-revenue branch: {top_branch} (${top_branch_revenue:,.2f})')
print(f'2. Highest-revenue product line: {top_product} (${top_product_revenue:,.2f})')
print(f'3. Highest-revenue customer type: {top_customer_type}')
print(f'4. Most-used payment method: {top_payment}')
print(f'5. Highest-revenue day of week: {top_day}')
```

## 12. Business Recommendations

Based on the results above, management could:

1. Focus promotional activity on the highest-performing product lines.
2. Investigate why lower-performing branches generate less revenue and identify successful practices from the leading branch.
3. Use customer-type and payment-method patterns to improve promotions and checkout experience.
4. Schedule inventory and staffing around the strongest sales days and hours.
5. Monitor customer ratings by branch to identify service-improvement opportunities.
